## Pandas 

- A package built on top of NumPy
- Two core objects in Pandas: DataFrame & the Series

```python
# DataFrame - a two-dimensional, size-mutable, tabular data structure with labeled axes (rows and columns)
# Basically like an excel table, but in Python
import pandas as pd

print("Example 1:")
display(pd.DataFrame({"Yes": [50, 21], "No": [131, 2]}))

print("Example 2:")
df = pd.DataFrame({
    "height": [170, 180, 175],
    "weight": [65, 80, 70]
})

display(df)
```

```python
# Change labels
pd.DataFrame({"Bob": ["I liked it.", "It was awful."], 
              "Sue": ["Pretty good.", "Bland."]},
             index=["Product A", "Product B"])
```

```python
# Series - is a sequence of data values; a single column of a DataFrame
# It is a one-dimensional labeled array capable of holding any data type 

display(pd.Series([1, 2, 3, 4, 5]))

# Assign row labels
display(pd.Series([30, 35, 40], index=['2015 Sales', '2016 Sales', '2017 Sales'], name='Product A'))

# You can think of DataFrame as a bunch of Series glued together.
```

```python
# Reading data files

# df = pd.read_csv("my_data.csv")
# pd.read_csv("my_data.csv", nrows=5) - read only first 5 rows of the csv file
# Has many other parameters to customize how the csv file is read, such as specifying the delimiter, handling missing values, etc.
    # index_col, index_row, header, names, skiprows, na_values, dtype, parse_dates, etc. 

data = pd.DataFrame({
    "height": [170, 180, 175],
    "weight": [65, 80, 70],
    "hair_color": ["black", "brown", "blonde"],
    "is_married": [False, True, False],
    "children": [0, 2, 1],
    "hobbies": [["reading", "hiking"], ["cooking", "traveling"], ["painting"]],
    "age": [30, 40, 35],
    "income": [50000, 80000, 60000]
})

print(data.shape) # (3, 6) - 3 rows, 6 columns

display(data.head(3)) # first 3 rows
display(data.tail(3)) # last 3 rows
```
